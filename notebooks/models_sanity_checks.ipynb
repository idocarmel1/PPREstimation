```python
import os
import sys

# # Standard interactive replacement for the 'parent directory' hack
# parent_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
# sys.path.append(parent_dir)

import pandas as pd
import numpy as np
from tqdm import tqdm

from PPRCalculator import PPRCalculator
from utils import species_groups, remove_cycles, mat_from_np, get_seq2name, get_DC, SpeciesGroup
```

```python
df = pd.DataFrame(species_groups)
model_numbers = list(df['model_number'].unique())

data = {
    'failed at loading': [],
    'broken models': [],
    'not balanced': [],
    'DC rows not sum to 1': [],
    'negative catch': [],
    'more than 1 DET seq': [],
    'less than 1 DET seq': [],
    'good models': [],
    'includes diet import': []
}

for model_number in tqdm(model_numbers):
    try:
        self = PPRCalculator(model_number)
        is_balanced, _, _ = self.is_model_balanced()
        DC_rows_sum_to_1 = bool(np.all(np.isclose(self._DC.sum(axis=1)[self.get_Regular_seq()], 1, atol=1e-4, rtol=1e-4)))
        catch_positive = all(self.catch >= 0)
        if not is_balanced:
            data['not balanced'].append(int(model_number))
        if not DC_rows_sum_to_1:
            data['DC rows not sum to 1'].append(int(model_number))
        if not catch_positive:
            data['negative catch'].append(int(model_number))
        if len(self.get_DET_seq()) > 1:
            data['more than 1 DET seq'].append(int(model_number))
        if len(self.get_DET_seq()) < 1:
            data['less than 1 DET seq'].append(int(model_number))
        if self.get_DC().loc[:, self.get_Import_seq()[0]].sum() > 0:
            data['includes diet import'].append(int(model_number))
        
        if not (is_balanced and DC_rows_sum_to_1 and catch_positive):
            data['broken models'].append(int(model_number))
        else:
            data['good models'].append(int(model_number))
    except:
        data['failed at loading'].append(int(model_number))
        continue

for k, v in data.items():
    print('='*20)
    print(f'{k} ({len(v)}): {v}')
```
